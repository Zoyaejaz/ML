# Types of ML on the basis of how it learns

Learning can either be  memorizing or generalizing. Memorizing means we learn or memorize it. generalizing means we understand the concept of it and not learning it.

Therefore Types of ML:
1. Instance Based
2. Model Based

# Instance Based

It is a machine learning approach where the system memorizes the training dataset as-is and makes predictions for new, unseen data by comparing it directly to the stored examples using a similarity metric.
there is no model to store here

# Model Based

Model-based learning is a machine learning paradigm where an AI builds an explicit mathematical or predictive model to generalize patterns from training data. Rather than memorizing specific examples, the algorithm uses the data to learn underlying rules, allowing it to accurately predict outcomes on completely new, unseen data.
here it store the model in suitable form.
