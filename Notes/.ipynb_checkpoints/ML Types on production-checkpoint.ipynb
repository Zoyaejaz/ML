# Types of ML on the basis of training on production

# BATCH/offline MACHINE LEARNING

This is a conventional way of training a ML model where we use entire dataset in one time to train the model, and not in a batch wise
Here, we pick an entire data, we train our machine learning model on an offline system. And then once the model is trained then we test the model and finally deploy it on the server.

# The problem with Batch learning

The biggest problem with this is that we have to retrain our model again and  again. Because data get updated periodically. The data get updated on different frequency(like weekly,monthly or yearly). 

So, what we have to do with this, we take the new updated data, old data and trained the model again on them and then test the model and at last we will deploy the model on the server.


Disadvantages of Batch ML
1. Lots of Data
2. Hardware Limitation
3. Availability

# Online Machine Learning

Here we train the model in batch wise.
It's called online learning because our model is getting trained when our model is on the server.
here we can train our model in the server only, when it faces a new type of data.
example: chatbot, youtube

In online learning, we perform incremental learning using mini batches of data where you train your model sequentially. 

When to use this learning?
1. Where there is a concept drift(concept drift means the place where concept get changes)
2. Cost Effective
3. Faster Solution
