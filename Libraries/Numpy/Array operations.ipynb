# Broadcasting

```python
l=[10,20,30,40]
arr=np.array(l)
arr
```

```python
for i in l:
    print(i+10)
```

```python
#to write the above code, in short way
arr+10                           #This is how broadcasting works(if i have single element and want that element to be added to each and every element of an array,we use broadcasting)
```

```python
#we can do this in the matrix also 
ar1=np.arange(1,26).reshape(5,5)
ar1
```

```python
ar1+10
```

# Deep and shallow Copy

```python
a=np.arange(1,21)
a
```

```python
slice=a[:5]
slice=slice*10
slice
```

```python
a   
# we see that on changing on the slice, we are not seeing any change in the original array.But when we do the same thing with the list,it will change the original array as well. 
#what array does, it makes another space in the ram.that's why changing on that array does not change the original array.
#this is a concept of shallow copy.
```

```python
#when on changing anything on the slice, will change the original array also, that concept is called deep copy.
```

```python
a
b=a
```

```python
b[0]=90
b
```

```python
a  #here we see the concept of deep copy. a also changes on changing b
```

```python
#a
#b=a
#if we want to make this code as shallow copy we can do by using .copy()
```

```python
a
b=a.copy()    #on using .copy(), it will follow the shallow copy.on changing b it will not change a
```

```python
b[0]=30
b
```

# STACKING ARRAY

Stacking means join two array. But we cannot stacked 2 matrix,because they are already stacked.
we can stack 2 arrays or we can stack 2 vectors
Remember, we have to use vectors and not matrix

```python
a=np.array([1,2,3,4])
b=np.array([5,6,7,8])
```

```python
#we can stack arrays in 2 ways
#1.vertical stack
#2.Horizontal stack
```

```python
np.vstack((a,b))    #This is vertically stack,here elements gets joint horizontally
```

```python
np.hstack((a,b))   #this is horizontally stack,here elements gets joint horizontally
```

```python
np.column_stack((a,b))  #This is a column stack
```

# Splitting Array

```python
c=np.arange(16).reshape(4,4)
c
```

```python
#we can split the arrays in 2 ways
#1.Horizontal split
#2.vertical split
```

```python
np.hsplit(c,2) 
#It will split c, horizontally in 2 parts 
#we can split the array in equal divisions. As there is 4*4 matrix, then we can split the array either in 2 or 4
```

```python
np.hsplit(c,3)   #this will give error, as 3 is no an equal division
```

```python
np.hsplit(c,4) 
```

```python
np.vsplit(c,2)  #this is vertically split. and same concept of equal divisions goes here
```
