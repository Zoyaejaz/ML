```python
import numpy as np
```

1. NumPy arrays are used to store large amounts of numbers.
2. An index is an element’s position in an array. NumPy arrays are 0-indexed.
3. Slicing is where we can access parts of a sequence with name[start:end].
4. 2D arrays have 1D arrays as elements.

# Numpy indexing and slicing of vectors

```python
arr=np.arange(11,21)
arr
```

#INDEXING

```python
arr[4]   #it will give 15 because in numpy,arrays are 0-based index
```

```python
arr[8]
```

#SLICING   --> It starts from the start index (inclusive) and ends before the end index (non-inclusive).

```python
arr[1:5]   #it will print numbers from index 1 to index 4. 
```

```python
arr[:5]  #here starting value automatically get set default to 0
```

```python
arr[:]  #it will print the entire array
```

```python
arr[3::2] #here 2 is the step size,it will print elements from 3 to 20 with step size of 2 between each and every element
```

# Numpy indexing and slicing of matrix

```python
arr=np.arange(1,31).reshape(6,5)
arr
```

#INDEXING

```python
arr[0] #it will print the entire 0th row
```

```python
arr[0,1]  #it will print the element present in 0th row and 1st column
```

```python
arr[5,4]
```

#SLICING

```python
arr[0:2,1:3]      #it will slice the row from 0th to 1st row and slice the column from 1st to 2nd column
```

```python
arr[3:6,3:5]
```

```python
# to select the column
arr[:,2]
```

```python
#Finding average
avg=np.average(arr)     #to find average we write it as np.average
avg
```

# Boolean Indexing

```python
arr=np.arange(11,21)
arr
```

```python
bool_index=arr%2==0     #if we want all even numbers present inside the matrix,we can do this instead of usig for loop
bool_index
#it will return true and false for each and every element
```

```python
arr=arr[bool_index]
arr
#here after passing the bool_index to array, it will give those values of matrix only which is giving true.
```
