# Matplotlib

```python
#Importing Matplotlib
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
```

# Matplotlib Basics

```python
x=np.linspace(0,5,11)     #it will generate the random 11 numbers from 0 to 5 with linear space in between all of them
x
```

```python
y=x**2
y
```

```python
plt.plot(x,y)    #plot make a line graph,in plot we have to pass axis, first one is x axis and 2nd one is y axis 
#vertical line is y axis and horizontal line is x axis

#Adding labels
plt.title("Square of a number")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
```

# Subplots in Matplotlib

Subplots means we want to create mutiple plots at one time.

```python
plt.subplot(2,2,1)   #subplot takes 3 arguments     #it will divide  the canvas into 2*2 box that's why we are seeing 4 boxes.And 1 means we are taking the 1st quadrant
plt.subplot(2,2,2)   #2 means we are taking the 2nd quadrant
plt.plot(x,y)       #it will plot the line for the just above code (in the 2nd quadrant)
plt.subplot(2,2,3)   #3 means we are taking the 3rd quadrant
plt.subplot(2,2,4)   #4 means we are taking the 4th quadrant
plt.plot(x**2,x**2)       #it will plot the line in the above code above plot(in the 4th quadrant)
```

# Objects Oriented way
