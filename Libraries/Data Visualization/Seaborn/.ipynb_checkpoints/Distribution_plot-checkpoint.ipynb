# Distribution plot

  we create distribution plots for quantitative data(like age,salary,etc.)

```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df=sns.load_dataset('tips')
df
```

```python
df['size'].unique()  # so we have only total_bill and tip only which are quantitative data
```

## Displot/histplot

```python
sns.displot(df['total_bill'], bins=10)  #either we can write distplot or histplot in this 
```

```python
#Kernal density estimator tells how much skewness your data is, it is written as kde
sns.displot(df['total_bill'],bins=20,kde=True)
```

```python
plt.subplot(1,2,1)
sns.histplot(df['total_bill'],bins=20,kde=True)
plt.subplot(1,2,2)
sns.histplot(df['tip'],bins=20,kde=True)
```

## Jointplot 
Joint plots combine a scatter plot with the distributions of the individual variables. This allows for a quick visual representation of how the variables are distributed individually and how they relate to one another.

```python
#It is used to compare two datas
```

```python
sns.jointplot(x='total_bill',y='tip',data=df,kind='scatter') #by default also,it gives you scatter,when you don't write kind in it
```

```python
sns.jointplot(x='total_bill',y='tip',data=df,kind='hex')
```

```python
sns.jointplot(x='total_bill',y='tip',data=df,kind='kde')
```

```python
sns.jointplot(x='total_bill',y='tip',data=df,kind='reg')  #reg means regression
```

## PairPlot
Pair plots are used explore relationships between several variables by generating scatter plots for every pair of variables in a dataset along with univariate distributions on the diagonal. This is useful for exploring datasets with multiple variables and seeing potential correlations.

```python
sns.pairplot(df)
```

```python
sns.pairplot(df,hue='sex',palette='rainbow') #hue tells in which category we are dividing the data. like in this we are dividing the data according to the gender i.e. sex wise
```
