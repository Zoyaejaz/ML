# Accessing Specific Column

Series: 
When we select a single column, we’re actually getting a Pandas Series (not a full DataFrame).
If a DataFrame is like a table, a Series is like a single column of that table. It’s still powerful, just narrower.

```python
import pandas as pd
characters_data = {
  'name': ['Thorne', 'Elira', 'Glim', 'Brug', 'Nyx', 'Kael', 'Mira', 'Drogan', 'Zara', 'Fenwick'],
  'race': ['Elf', 'Human', 'Gnome', 'Half-Orc', 'Tiefling', 'Dragonborn', 'Halfling', 'Dwarf', 'Aasimar', 'Goblin'],
  'class': ['Ranger', 'Cleric', 'Wizard', 'Barbarian', 'Rogue', 'Paladin', 'Bard', 'Fighter', 'Sorcerer', 'Warlock'],
  'level': [5, 3, 4, 2, 6, 7, 3, 5, 4, 2],
  'hp': [42, 28, 33, 25, 48, 56, 30, 44, 36, 24],
  'alignment': [
    'Chaotic Good', 'Lawful Good', 'Neutral', 'Chaotic Neutral', 'Chaotic Evil',
    'Lawful Neutral', 'Neutral Good', 'Neutral', 'Chaotic Good', 'Lawful Evil'
  ]
}
characters=pd.DataFrame(characters_data)
```

```python
characters
```

```python
#Accessing Single column
characters_name=characters['name']
characters_name
```

```python
type(characters_name)
```

```python
type(characters)
```

```python
#Accessing Multiple Columns
df=characters[['race','level']]
df
```

```python
type(df)  #It gives dataframe because it is selecting more than one column and not a single column 
```

```python
#Accessing all but one column with .drop()
remove_alignment_df=characters.drop("alignment",axis=1)
remove_alignment_df
```

# Find Missing values

```python
import numpy as np
import pandas as pd
```

```python
data={
    'A':[1,2,np.nan,4,5],
    'B':[np.nan,2,3,4,5],
    'C':[1,2,3,np.nan,np.nan],
    'D':[1,np.nan,np.nan,np.nan,5]
}
df=pd.DataFrame(data)
```

```python
df
```

```python
df.isna()
```
