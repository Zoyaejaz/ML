```python
import numpy as np
import pandas as pd
```

```python
data = {
    'Date': pd.date_range('2023-01-01', periods=20),
    'Product': ['A', 'B', 'C', 'D'] * 5,
    'Region': ['East', 'West', 'North', 'South', 'East', 'West', 'North', 'South', 'East', 'West',
               'North', 'South', 'East', 'West', 'North', 'South', 'East', 'West', 'North', 'South'],
    'Sales': np.random.randint(100, 1000, 20),
    'Units': np.random.randint(10, 100, 20),
    'Rep': ['John', 'Mary', 'Bob', 'Alice', 'John', 'Mary', 'Bob', 'Alice', 'John', 'Mary',
            'Bob', 'Alice', 'John', 'Mary', 'Bob', 'Alice', 'John', 'Mary', 'Bob', 'Alice']
}

df = pd.DataFrame(data)
```

```python
df['Month']=df['Date'].dt.month_name()
df['Quarter']='Q'+df['Date'].dt.quarter.astype(str)
df
```

# Pivot tables

```python
pd.pivot_table(df,values='Sales',index='Region',columns='Product')   #using pivot,we can create our own table by creating its index,column and data all by our own
```

#we get NaN for the empty data which is not there in the table and if any index has more than 1 value then we take the mean in that case and write that in the index. here we get mean because the aggregation function in that is set as mean by default, we can change it by changing it to other function.

```python
pivot2=pd.pivot_table(df,values=['Sales','Units'],index='Region',columns='Product')
pivot2
```

```python
pivot3=pd.pivot_table(df,values=['Units','Sales'],index='Date',columns='Rep')
pivot3

#for values, we can only use the numerical value column because the aggregation function works on the numerical value only.
```

# Cross Tabs

```python
pd.crosstab(df['Region'],df['Product'])   
```

In pivot_table, it gives mean values, in crosstab it gives count for it
Crosstab is used for counting purpose and pivot_table is used for aggregation purpose.
