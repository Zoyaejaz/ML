```python
import numpy as np
import pandas as pd
```

```python
data = {
    'Category': ['A', 'B', 'A', 'B', 'A', 'B', 'A', 'B'],
    'Store': ['S1', 'S1', 'S2', 'S2', 'S1', 'S2', 'S2', 'S1'],
    'Sales': [100, 200, 150, 250, 120, 180, 200, 300],
    'Quantity': [10, 15, 12, 18, 8, 20, 15, 25],
    'Date': pd.date_range('2023-01-01', periods=8)
}
df = pd.DataFrame(data)
```

```python
df
```

# Groupby

```python
# Group by category and calculate the sum of sales
cat=df.groupby('Category')
cat
```

```python
for i,v in cat:
    print(i)
    print(v)
```

```python
cat=df.groupby('Category')['Sales'].sum()
cat
```

```python
# Group by Stores and calculates the sum of sales
st=df.groupby('Store')['Sales'].sum()
st
```

Types of data
1. Qualitative Data: It means categorical data. eg:gender,categoryetc.
2. Quantitative Data:It contains numeric data like age, name sales etc.

Groupby method is done for category data type

```python
#Group by multiple columns
#Group by Category and Store
cat=df.groupby(['Category','Store'])['Sales'].sum()
cat
```
