# Changing Data Types

After finding the data types for each and every column, now we can change its datatype

#If Pandas is treating a numeric column as a string, we can use pd.to_numeric() to convert the data type.
    
    ------>  example: df['age'] = pd.to_numeric(df['age'])

--> This will change the age column to be numeric. But this only works if every value in the age column can be interpreted as a number.
If one of the values is something like 'twenty years old', Pandas won't know how to turn that into 20. In this case, pd.to_numeric will throw an error!

To solve this, we can use the parameter errors = 'coerce'

example: df['age'] = pd.to_numeric(df['age'], errors = 'coerce')

-->When using this parameter, rather than throwing an error for values that it can not convert, Pandas will convert them to NaN (not a number).




```python
# convert the appropriate columns to numeric types, and then find the average (.mean()) high score and time played.
score_Card['score']=pd.to_numeric(score_Card['score'],errors='coerce')
score_Card['level']=pd.to_numeric(score_Card['level'],errors='coerce')
score_Card['time_solved']=pd.to_numeric(score_Card['time_solved'],errors='coerce')
score_Card.dtypes
```

```python
#finding the mean of numeric column
score_Card['score'].mean()
```
