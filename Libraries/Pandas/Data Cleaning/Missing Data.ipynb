# 2. HANDLE MISSING DATA

###1.  Remove Rows with .dropna()

---> This removes any rows that contain at least one missing value. If only a small number of rows are missing data (and that data isn’t important), we can safely drop those rows with .dropna() which stands for "Drop Not Available.
        cleaned_df = df.dropna()

But what if we only care about missing values in specific columns? We can use subset
        cleaned_df = df.dropna(subset = ['username', 'location'])
    -->This removes rows that have a missing value in the username or location columns.

```python
#Create a DataFrame that removes rows with missing data in any column.
given_data=addresses.dropna()
given_data
```

```python
# Create a DataFrame that removes rows with missing data in only the class column.
cleaned=addresses.dropna(subset=['zip_code','city'])
cleaned
```

###2.Replace Missing Values with .fillna()

-->Imputation means filling in the blanks with a value that makes sense. Your choice on how to do this will depend on the data. Some common choices include:

        1.A default value, like 'Unknown'
            example: df['location'] = df['location'].fillna('Unknown')
            
        
        2.The most common value (the mode)
           example:        # Find the most common value in the column
                             most_common_location = df['location'].mode()[0]
                           # Fill missing values with that value
                             df['location'] = df['location'].fillna(most_common_location)
                             
                             
        3.The average value (if your data is numerical)
            example:        mean_age = df['age'].mean()
                            df['age'] = df['age'].fillna(mean_age)


```python
# Impute the missing zip_code with the average... and hope the mailman delivers it. 
mean_zip=addresses['zip_code'].mean()
addresses['zip_code']=addresses['zip_code'].fillna(mean_zip)
```

```python
# Impute the missing state and delivery_type values with the most common values.
most_common=addresses['delivery_type'].mode()[0]
addresses['delivery_type']=addresses['delivery_type'].fillna(most_common)
addresses
```
