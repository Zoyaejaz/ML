# Find Duplicate Data with .duplicated()

We can use the .duplicated() method to find rows that are duplicates of each other.
Syntax:  df.duplicated() ---> It returns a Series of True and False values. True means that row is a duplicate

If we wanted to see the actual rows that are duplicates, we could then use this Series as a filter to our DataFrame
Syntax:  df[df.duplicated()] ---> This will return only the rows from the Series that were marked as True.


```python
# Find the duplicate from the given data
duplicate=df.duplicated()
duplicate
```

```python
# To see the actual rows that are duplicates
actual_duplicate=df[df.duplicated()]
actual_duplicate
```

# Remove Duplicates with .drop_duplicates()

Once we know if we have duplicates, we can decide how to remove them. The simplest way is to use .drop_duplicates(). This will remove all rows that are exactly identical to another row in the DataFrame.

Syntax:   no_dupes = df.drop_duplicates()

However, sometimes rows aren’t exact duplicates, but you still want to remove certain repeated values.
For example, in a library catalog, the same book might be entered twice with slightly different author names. The Lord of the Rings might have "JRR Tolkien" as the author while another row might use "J.R.R. Tolkien".
These rows aren’t exact duplicates, so .drop_duplicates() on the whole row won’t work.
Luckily, most datasets have a column that uniquely identifies each row. For a library, this would be the book’s ISBN.

We can tell Pandas to remove duplicates based only on that column using subset:

#Removes rows that have a duplicate ISBN with another row

unique_books = catalog.drop_duplicates(subset = ['ISBN'])   --->Pandas will keep the first row for each unique ISBN and will remove all other rows with duplicated ISBNs.

```python
# create a DataFrame with one row for each unique ticket booking
no_duplicate=df.drop_duplicates()
no_duplicate
```
