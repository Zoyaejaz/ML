# String Manipulation

There are a handful of very helpful .str methods for basic string processing:

.str.lower(): makes everything lowercase

.str.upper(): makes everything uppercase

.str.strip(): removes whitespace from both ends

.str.contains('text'): checks for a substring

.str.replace('old', 'new'): replaces parts of the text

You can even chain these methods together in a single line

```python
import pandas as pd
import random

random.seed(1)

# Different possible RSVP variants
options = ['Yes', 'yes', 'yesss', 'Yep   ', ' yes', 'YES', 'y', 'No  ', 'no', 'N ', 'n', 'no thx']

one_hundred_responses = random.choices(options, k=100)

# Generate fake names for the 100 guests
names = [f'Guest_{i+1}' for i in range(100)]

# Create the DataFrame
responses = pd.DataFrame({
  'name': names,
  'rsvp': one_hundred_responses
})
```

```python
responses
```

#our job is to process the text of the rsvp column and make each response either 'y' or 'n'
1. Begin by looking at the head or tail of the DataFrame to understand what the data looks like.
2. Then, standardize the capitalization and strip extra spaces of the rsvp column.
3. Finally, use .replace() to turn yes into y and no into n.
4.We can use responses['rsvp'].value_counts() to print out the final number of yes and no responses.

```python
#1
responses.head()
```

```python
responses.tail()
```

```python
#2
responses['rsvp']=responses['rsvp'].str.lower().str.strip()
```

```python
responses.head(10)
```

```python
#3
responses['rsvp']=responses['rsvp'].str.replace('yes','y')
responses['rsvp']=responses['rsvp'].str.replace('yss','y')
responses['rsvp']=responses['rsvp'].str.replace('yep','y')
```

```python
responses['rsvp']=responses['rsvp'].str.replace('no','n')
responses['rsvp']=responses['rsvp'].str.replace('n thx','n')
```

```python
responses['rsvp'].value_counts()
```
